```python
import pandas as pd

# Загружаем данные

data = pd.read_csv('data/train.csv')
data = data.drop_duplicates()

# Удаляем целевой признак

X = data.drop('Is_Bestseller', axis=1)

# Сэмплируем 500 случайных строк

sample = X.sample(n=500, random_state=42)

# Сохраняем

sample.to_csv('data/test_sample.csv', index=False)
```
